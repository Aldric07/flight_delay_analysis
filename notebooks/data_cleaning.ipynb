# ✈️ Airline Delay Analysis
Analyse des retards aériens par compagnie et par aéroport aux États-Unis.

```python
import pandas as pd

df = pd.read_csv("../data/Airline_Delay_Cause.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()

df["year"] = df["year"].astype(int)
df["month"] = df["month"].astype(int)

df = df[df["arr_flights"] > 0]
```

```python
df["delay_rate"] = df["arr_del15"] / df["arr_flights"]
df["cancel_rate"] = df["arr_cancelled"] / df["arr_flights"]

df["total_delay_minutes"] = (
    df["carrier_delay"] +
    df["weather_delay"] +
    df["nas_delay"] +
    df["security_delay"] +
    df["late_aircraft_delay"]
)

df["period"] = pd.to_datetime(df["year"].astype(str) + "-" + df["month"].astype(str) + "-01")
```

```python
df[["delay_rate", "cancel_rate", "total_delay_minutes"]].describe()
```

```python
df.to_csv("../data/airline_delay_clean.csv", index=False)
```
